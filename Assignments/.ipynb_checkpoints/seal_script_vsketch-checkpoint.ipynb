```
import vsketch
from bezmerizing import Path
from shapely.geometry import LineString
import random
```

```
class SealScriptSketch(vsketch.SketchClass):
    def draw(self):
        self.size("100mm", "100mm")
        self.scale("mm")

        self.stroke(1)
        self.stroke_width(0.4)
        self.no_fill()

        # Outer seal border
        self.rect(10, 10, 80, 80)

        # Generate a grid of fake seal characters
        cols, rows = 4, 4
        cell_w = 80 / cols
        cell_h = 80 / rows
        margin_x = 10
        margin_y = 10

        for row in range(rows):
            for col in range(cols):
                cx = margin_x + col * cell_w
                cy = margin_y + row * cell_h
                self.character(cx, cy, cell_w, cell_h)

    def character(self, x, y, w, h):
        # Draw a pseudo seal-script character using geometric strokes
        num_strokes = random.randint(3, 6)
        for _ in range(num_strokes):
            sx = x + random.uniform(w * 0.1, w * 0.9)
            sy = y + random.uniform(h * 0.1, h * 0.9)
            ex = x + random.uniform(w * 0.1, w * 0.9)
            ey = y + random.uniform(h * 0.1, h * 0.9)
            p = Path(LineString([(sx, sy), (ex, ey)]))
            self.vs.geometry(p)
```

```
SealScriptSketch.display()
```
