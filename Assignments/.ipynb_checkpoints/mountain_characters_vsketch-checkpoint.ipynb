# Mountain Characters with Vsketch

This notebook generates character-like visuals inspired by Chinese traditional mountain landscape painting (中国传统山水国画).

```python
import vsketch
import numpy as np
import random
```

```python
class MountainCharacterSketch(vsketch.SketchClass):
    def draw(self, vsk: vsketch.Vsketch) -> None:
        vsk.size("210mm", "297mm")  # A4 size
        vsk.scale("mm")
        vsk.penWidth(0.2)
        
        cols = 4
        rows = 6
        spacing_x = 40
        spacing_y = 40

        for row in range(rows):
            for col in range(cols):
                x = 20 + col * spacing_x
                y = 30 + row * spacing_y
                self.draw_mountain_char(vsk, x, y)

    def draw_mountain_char(self, vsk: vsketch.Vsketch, cx: float, cy: float):
        peaks = random.randint(3, 6)
        width = random.uniform(10, 20)
        height = random.uniform(10, 25)

        # Base line
        points = []
        for i in range(peaks):
            px = cx - width/2 + (i / (peaks - 1)) * width
            py = cy - random.uniform(0, height)
            points.append((px, py))

        # Optional: add small vertical strokes to resemble ancient text
        for px, py in points:
            if random.random() < 0.5:
                vsk.line(px, py, px, py - random.uniform(2, 5))

        # Connect with curves to create mountain shape
        for i in range(len(points) - 1):
            x0, y0 = points[i]
            x1, y1 = points[i + 1]
            ctrl_x = (x0 + x1) / 2 + random.uniform(-2, 2)
            ctrl_y = min(y0, y1) - random.uniform(2, 5)
            vsk.curve(x0, y0, ctrl_x, ctrl_y, x1, y1)
```

```python
MountainCharacterSketch.display()
```
