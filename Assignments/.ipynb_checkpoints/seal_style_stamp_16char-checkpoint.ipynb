# 篆刻风格印章生成器（Seal-style Stamp Generator）

This notebook uses `vsketch` to create a visual imitation of a traditional Chinese seal (篆刻印章), laid out as a 4x4 grid of abstract glyphs or characters.

```python
import vsketch
import numpy as np
import random
```

```python
class SealStyleSketch(vsketch.SketchClass):
    def draw(self, vsk: vsketch.Vsketch) -> None:
        vsk.size("100mm", "100mm")
        vsk.scale("mm")
        vsk.penWidth(0.3)
        
        # Optional border
        margin = 5
        seal_size = 90
        vsk.rect(margin, margin, seal_size, seal_size)

        cell_size = seal_size / 4

        for row in range(4):
            for col in range(4):
                cx = margin + col * cell_size + cell_size / 2
                cy = margin + row * cell_size + cell_size / 2
                self.draw_seal_char(vsk, cx, cy, cell_size * 0.8)

    def draw_seal_char(self, vsk: vsketch.Vsketch, cx: float, cy: float, size: float):
        num_strokes = random.randint(3, 6)
        for _ in range(num_strokes):
            x0 = cx + random.uniform(-size/2, size/2)
            y0 = cy + random.uniform(-size/2, size/2)
            if random.random() < 0.5:
                # horizontal or vertical
                if random.random() < 0.5:
                    x1 = x0
                    y1 = cy + random.uniform(-size/2, size/2)
                else:
                    x1 = cx + random.uniform(-size/2, size/2)
                    y1 = y0
            else:
                # diagonal
                x1 = cx + random.uniform(-size/2, size/2)
                y1 = cy + random.uniform(-size/2, size/2)
            vsk.line(x0, y0, x1, y1)
```

```python
SealStyleSketch.display()
```
