# 篆刻风格印章（Seal Style Block Grid, No Red）

This notebook creates a 4×4 grid of stylized characters imitating a Chinese seal (篆刻印章) in layout and geometry—but using black ink instead of red.

```python
import vsketch
import random
```

```python
class GeometricSealSketch(vsketch.SketchClass):
    def draw(self, vsk: vsketch.Vsketch) -> None:
        vsk.size("100mm", "100mm")
        vsk.scale("mm")
        vsk.penWidth(0.5)

        margin = 5
        grid_size = 90
        cell_size = grid_size / 4

        # Outer border
        vsk.rect(margin, margin, grid_size, grid_size)

        # Create 4x4 block layout
        for row in range(4):
            for col in range(4):
                x = margin + col * cell_size
                y = margin + row * cell_size
                self.draw_block_char(vsk, x, y, cell_size)

    def draw_block_char(self, vsk: vsketch.Vsketch, x: float, y: float, size: float):
        pad = size * 0.1
        line_width = size * 0.08
        grid_unit = (size - 2 * pad) / 3

        # Draw a character-like pattern based on a 3x3 grid
        for i in range(3):
            for j in range(3):
                if random.random() < 0.6:
                    x0 = x + pad + i * grid_unit
                    y0 = y + pad + j * grid_unit
                    vsk.rect(x0, y0, grid_unit, grid_unit)
```

```python
GeometricSealSketch.display()
```
