# Refined Mountain Characters with Classical Chinese Style

This notebook generates character-like mountain formations inspired by traditional Chinese landscape painting (山水画), with soft layering, textural strokes, and misty contours.

```python
import vsketch
import numpy as np
import random
```

```python
class ClassicalMountainCharacterSketch(vsketch.SketchClass):
    def draw(self, vsk: vsketch.Vsketch) -> None:
        vsk.size("210mm", "297mm")
        vsk.scale("mm")
        vsk.penWidth(0.2)

        cols = 4
        rows = 5
        spacing_x = 45
        spacing_y = 45

        for row in range(rows):
            for col in range(cols):
                x = 25 + col * spacing_x
                y = 35 + row * spacing_y
                self.draw_mountain_char(vsk, x, y)

    def draw_mountain_char(self, vsk: vsketch.Vsketch, cx: float, cy: float):
        # Generate layered peaks to mimic Chinese mountain structure
        layers = random.randint(2, 4)
        width = random.uniform(12, 20)
        total_height = random.uniform(15, 30)
        layer_gap = total_height / layers

        for layer in range(layers):
            base_y = cy - (layer * layer_gap)
            height = random.uniform(4, 8)
            peaks = random.randint(3, 5)
            points = []

            for i in range(peaks):
                px = cx - width / 2 + (i / (peaks - 1)) * width
                py = base_y - random.uniform(0, height)
                points.append((px, py))

            for px, py in points:
                # Vertical texture strokes like moss dots or tree brush
                if random.random() < 0.6:
                    vsk.line(px, py, px, py - random.uniform(1, 3))

            for i in range(len(points) - 1):
                x0, y0 = points[i]
                x1, y1 = points[i + 1]
                ctrl_x = (x0 + x1) / 2 + random.uniform(-1.5, 1.5)
                ctrl_y = min(y0, y1) - random.uniform(1, 3)
                vsk.bezier(x0, y0, ctrl_x, ctrl_y, ctrl_x, ctrl_y, x1, y1)

        # Add a mist effect with faint wavy line
        if random.random() < 0.8:
            mist_base = cy + random.uniform(1, 3)
            for i in range(6):
                mx0 = cx - width/2 + i * (width / 5)
                mx1 = mx0 + (width / 5)
                my = mist_base + np.sin(i + random.uniform(0, 0.5)) * 1.2
                vsk.line(mx0, my, mx1, my)
```

```python
ClassicalMountainCharacterSketch.display()
```
