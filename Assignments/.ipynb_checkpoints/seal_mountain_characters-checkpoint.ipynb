# 篆刻风格的山形字符

This notebook generates a 4x4 grid of mountain-shaped characters, imitating the style of ancient Chinese seal carving (篆刻).

```python
import vsketch
import numpy as np
import random
```

```python
class SealMountainCharacters(vsketch.SketchClass):
    def draw(self, vsk: vsketch.Vsketch) -> None:
        vsk.size("80mm", "80mm")  # Square seal format
        vsk.scale("mm")
        vsk.penWidth(0.25)
        vsk.stroke(1)

        cols = 4
        rows = 4
        spacing = 18
        margin = 8

        for row in range(rows):
            for col in range(cols):
                x = margin + col * spacing
                y = margin + row * spacing
                self.draw_character(vsk, x, y)

        # Seal border
        vsk.rect(margin - 3, margin - 3, spacing * cols + 6, spacing * rows + 6)

    def draw_character(self, vsk: vsketch.Vsketch, cx: float, cy: float):
        layers = random.randint(1, 3)
        width = random.uniform(8, 12)
        total_height = random.uniform(8, 14)
        layer_gap = total_height / layers

        for layer in range(layers):
            base_y = cy - total_height / 2 + layer * layer_gap
            height = random.uniform(2, 5)
            peaks = random.randint(2, 4)
            points = []

            for i in range(peaks):
                px = cx - width/2 + (i / (peaks - 1)) * width
                py = base_y - random.uniform(0, height)
                points.append((px, py))

            for px, py in points:
                if random.random() < 0.5:
                    vsk.line(px, py, px, py - random.uniform(1, 2))

            for i in range(len(points) - 1):
                x0, y0 = points[i]
                x1, y1 = points[i + 1]
                ctrl_x = (x0 + x1) / 2 + random.uniform(-1, 1)
                ctrl_y = min(y0, y1) - random.uniform(1, 2)
                vsk.bezier(x0, y0, ctrl_x, ctrl_y, ctrl_x, ctrl_y, x1, y1)
```

```python
SealMountainCharacters.display()
```
